## Persona?
* ***Prompt Engineering***
* Giving identity to the chatbot!
* This document uses `ChatPromptTemplate`'s `SystemMessage`.
* The contents in `SystemMessage` will let the chatbot remember its  identity whenever we ask questions.

```python
from langchain_openai import ChatOpenAI
```

```python
from langchain.prompts import ChatPromptTemplate
from langchain_core.messages import SystemMessage
from langchain.prompts import HumanMessagePromptTemplate
```

```python
from IPython.display import Markdown, display
def printmd(string):
    display(Markdown(string))
```

```python
question1 = "합성의약품에 대해서 설명해주세요."
question2 = "바이오의약품에 대해서 설명해주세요."
```

```python
model = ChatOpenAI(
    model_name = "gpt-4o-mini",
    temperature = 0
)
```

### POV: Pharmacy professor to college students

```python
chat_template1 = ChatPromptTemplate.from_messages(
    [
        SystemMessage(
            content = "당신은 약학대학 교수입니다. 당신의 청중은 대학생입니다."
        ),
        HumanMessagePromptTemplate.from_template("{text}")
    ]
)
```

```python
### question 1
message1 = chat_template1.format_messages(text=question1)
```

```python
message1
```

```python
out1 = model.invoke(message1)
```

```python
printmd(out1.content)
```

```python
### question 2
message2 = chat_template1.format_messages(text=question2)
out2 = model.invoke(message2)
```

```python
printmd(out2.content)
```

```python
chat_template3 = ChatPromptTemplate.from_messages(
    [
        SystemMessage(
            content = "당신은 생명과학대 교수입니다. 당신의 청중은 화학과 교수님들입니다."
        ),
        HumanMessagePromptTemplate.from_template("{text}")
    ]
)
message2 = chat_template3.format_messages(text=question2)
out2 = model.invoke(message2)
printmd(out2.content)
```
