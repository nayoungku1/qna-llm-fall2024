## Output Parser
* 다른 작업을 위해 내가 원하는 답은 중간답 (예, 답이 "케이팝 아티스트 - 그에 대한 설명"일 때 아티스트명)만 받고 싶을 때
* 따라서 natural language의 형태가 아닌 내가 원하는 형태의 답을 받고자 할 때 사용

```python
from langchain_openai import ChatOpenAI
from langchain.chains import LLMChain
from langchain.prompts import PromptTemplate
```

```python
from IPython.display import Markdown, display
def printmd(string):
    display(Markdown(string))
```

```python
model = ChatOpenAI(
    model_name = "gpt-4o-mini",
    temperature=0
)
```

### Default

```python
prompt1 = PromptTemplate(
    template = "Give me {how_many} famous K-pop artists.",
    input_variables = ["how_many"]
)
```

```python
prompt1
```

```python
chain1 = prompt1 | model
```

```python
out1 = chain1.invoke(
    {
        "how_many" : 3
    }
)
```

```python
type(out1)
```

```python
printmd(out1.content)
```

```python
type(out1.content)
```

### output parser의 적용

```python
from langchain.output_parsers import CommaSeparatedListOutputParser
```

```python
output_parser = CommaSeparatedListOutputParser()
format_instructions = output_parser.get_format_instructions()
```

```python
format_instructions
```

```python
prompt2 = PromptTemplate(
    template = "Give me {how_many} famous K-pop artists.\n{format_instructions}",
    input_variables = ["how_many"],
    partial_variables = {"format_instructions": format_instructions}
)
```

```python
chain2 = prompt2 | model | output_parser
```

```python
out2 = chain2.invoke(
    {
        "how_many": 3
    }
)
```

```python
out2
```

```python
type(out2)
```
