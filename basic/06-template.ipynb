```python
from langchain_openai import ChatOpenAI
```

```python
model = ChatOpenAI(
    model_name="gpt-4o-mini",
    temperature=0
)
```

```python
from langchain.prompts import PromptTemplate
```

```python
prompt = PromptTemplate.from_template(
    """
    You are the one of the most famous {kind} cuisine chef.
    Using the input ingredient down below, make one recipe.
    Please answer the name of the food first, the ingredient you utilized, and then lastly the step for cooking.
    <Ingredients>
    {ingredients}
    """
)
```

```python
prompt
```

### Italian cuisine

```python
prompt_message = prompt.format(kind="Italian", ingredients="mushrooms, parsley, noodles, green peppers, cucumber,onion")
```

```python
print(prompt_message)
```

```python
for chunk in model.stream(prompt_message):
    print(chunk.content, end="", flush=True)
```
