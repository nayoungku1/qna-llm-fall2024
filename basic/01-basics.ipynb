### Basics

```python
from openai import OpenAI
```

```python
client = OpenAI()
```

```python
client
```

```python
out = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "user",
            "content": "Who won the F1 Monaco Grand Prix in 2024?"
        }
    ]
)
```

```python
out
```

```python
out.choices[0].message.content
```

### using invoke with langchain package

```python
from langchain_openai import ChatOpenAI
```

```python
chat = ChatOpenAI(
    model_name="gpt-4o-mini"
)
```

```python
out = chat.invoke("Who won the F1 Monaco Grand Prix in 2024?")
```

```python
out #AIMessage format
```

```python
out.content
```
