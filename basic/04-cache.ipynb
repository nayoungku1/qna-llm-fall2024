## Cache
* If we set the cache, it search on the history.
* Then if there are same question and its answer, it brings the answer.
* Time efficiency!

```python
from langchain_openai import ChatOpenAI
```

```python
from IPython.display import Markdown, display
def printmd(string):
    display(Markdown(string))
```

```python
### Cache setting:

from langchain.cache import InMemoryCache
from langchain.globals import set_llm_cache

set_llm_cache(InMemoryCache())
```

```python
question = "화석에 대해서 설명해주세요."
```

```python
model = ChatOpenAI(
    model_name = "gpt-4o-mini",
    temperature = 0
)
```

```python
%%time

out = model.invoke(question)
```

```python
printmd(out.content)
```

```python
%%time

out = model.invoke(question)
```

4.29 s -> 788 µs !!!
