## Streaming?
* When we chat in ChatGPT, the AI answer shows up on the screen as if it is typing.
* Not cumulating the answers and printing the whole thing once, but printing chunks many times
* **Advantage**: Waiting for the answer would be pretty boring. It is even more for the question requiring long answer. So, it is helpful for clients' patience!

```python
from langchain_openai import ChatOpenAI
```

```python
model = ChatOpenAI(
    model_name="gpt-4o-mini"
)
```

```python
question = "화석에 대해서 설명해주세요."
```

```python
for chunk in model.stream(question):
    print(chunk.content, end="", flush=True)
```
