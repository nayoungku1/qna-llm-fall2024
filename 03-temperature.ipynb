## Temperature?
* Temperature is about the reproducibility of the codes.
* If the task requires creativity (like writing a fantasy novel,) it is better to set the temp = 1
* If the task requires stable answer, temp = 0 is better. 

```python
from langchain_openai import ChatOpenAI
```

```python
from IPython.display import Markdown, display
def printmd(string):
    display(Markdown(string))
```

```python
question = "화석에 대해서 설명해주세요."
```

### Case 01. temp == 0

```python
chat1 = ChatOpenAI(
    model_name = "gpt-4o-mini",
    temperature = 0
)
```

```python
# trial 01
out = chat1.invoke(question)
printmd(out.content)
```

```python
# trial 02
out = chat1.invoke(question)
printmd(out.content)
```

### Case 02. temp == 1

```python
chat2 = ChatOpenAI(
    model_name = "gpt-4o-mini",
    temperature = 1
)
```

```python
# trial 01
out = chat2.invoke(question)
printmd(out.content)
```

```python
# trial 02
out = chat2.invoke(question)
printmd(out.content)
```
