## Similarity Search
* 저장된 자료와 질문의 유사도를 비교하여, 가장 비슷한 자료를 꺼내오기 위해서 이러한 방법을 채택한다.

```python
sentence = [
    "안녕하세요",
    "안녕",
    "잘가",
    "잘자",
    "반갑다"
]
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def get_similarity(a,b):
    return cosine_similarity([a], [b])[0][0]
```

```python
import pandas as pd

def compare(embeddings):
    out = []
    for i, emb1 in enumerate(embeddings):
        for j, emb2 in enumerate(embeddings):
            if i < j:
                score = get_similarity(emb1, emb2)
                sentence1, sentence2 = sentence[i], sentence[j]
                out.append((sentence1, sentence2, score))

    return pd.DataFrame(out, columns=["sentence1", "sentence2", "similarity_score"])
```

```python
from langchain_openai import OpenAIEmbeddings
```

```python
embedding_model = OpenAIEmbeddings(model="text-embedding-3-small")
```

```python
embeddings = embedding_model.embed_documents(sentence)
```

```python
compare(embeddings)
```

* 임베딩모델에 따라 다른 결과가 나온다.
* 답 != 백터디비에서 유사도 검색
