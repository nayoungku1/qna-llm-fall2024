```python
from langchain_openai import ChatOpenAI
```

```python
from IPython.display import Markdown, display
def printmd(string):
    display(Markdown(string))
```

```python
model = ChatOpenAI(
    model_name = "gpt-4o-mini",
    temperature=0
)
```

```python
question = "삼엽충 화석은 진화론을 뒷받침하고 있나요?"
```

```python
out = model.invoke(question)
```

```python
printmd(out.content)
```
